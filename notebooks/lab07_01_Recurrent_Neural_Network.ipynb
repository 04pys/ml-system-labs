```python
# RNN 구현
```

```python
import numpy as np
```

```python
timesteps = 10 # 시점의 수. NLP에서는 보통 문장의 길이가 됨
input_size = 4 # 입력의 차원. NLP에서는 보통 단어 벡터 차원이 된다
hidden_size = 8 # 은닉 상태의 크기. 히든 셀의 용량이다
```

```python
inputs = np.random.random((timesteps, input_size)) # 입력에 해당되는 2D 텐서
```

```python
hidden_state_t = np.zeros((hidden_size,)) # 초기 은닉 상태는 0(벡터)로 초기화
# 은닉 상태의 크기 hidden_size로 은닉 상태를 만듬.
```

```python
print(hidden_state_t) # 8의 크기를 가지는 은닉 상태. 현재는 초기 은닉 상태로 모든 차원이 0의 값을 가짐.
```

```python
Wx = np.random.random((hidden_size, input_size))  # (8, 4)크기의 2D 텐서 생성. 입력에 대한 가중치.
Wh = np.random.random((hidden_size, hidden_size)) # (8, 8)크기의 2D 텐서 생성. 은닉 상태에 대한 가중치.
b = np.random.random((hidden_size,)) # (8,)크기의 1D 텐서 생성. 이 값은 편향(bias).
```

```python
print(np.shape(Wx))
print(np.shape(Wh))
print(np.shape(b))
```

```python
total_hidden_states = []

# 메모리 셀 동작
for input_t in inputs: # 각 시점에 따라서 입력값이 입력됨.
  output_t = np.tanh(np.dot(Wx,input_t) + np.dot(Wh,hidden_state_t) + b) # Wx * Xt + Wh * Ht-1 + b(bias)
  total_hidden_states.append(list(output_t)) # 각 시점의 은닉 상태의 값을 계속해서 축적
  print(np.shape(total_hidden_states)) # 각 시점 t별 메모리 셀의 출력의 크기는 (timestep, output_dim)
  hidden_state = output_t

total_hidden_states = np.stack(total_hidden_states, axis = 0)
# 출력 시 값을 깔끔하게 해준다.

print(total_hidden_states) # (timesteps, output_dim)의 크기. 이 경우 (10, 8)의 크기를 가지는 메모리 셀의 2D 텐서를 출력.
```

```python
# pytorch로 RNN구현
```

```python
import torch
import torch.nn as nn
```

```python
input_size = 5 # 입력의 크기. 매 timestep마다 들어가는 입력의 크기
hidden_size = 8 # 은닉 상태의 크기
```

```python
# (batch_size, time_steps, input_size)
inputs = torch.Tensor(1, 10, 5)
```

```python
cell = nn.RNN(input_size, hidden_size, batch_first=True)
```

```python
outputs, _status = cell(inputs)
```

```python
print(outputs.shape) # 모든 time-step의 hidden_state
```

```python
print(_status.shape) # 최종 time-step의 hidden_state
```

```python
# 층이 2개인 Deep Recurrent Neural Network 구현
```

```python
# (batch_size, time_steps, input_size)
inputs = torch.Tensor(1, 10, 5)
```

```python
cell = nn.RNN(input_size = 5, hidden_size = 8, num_layers = 2, batch_first=True)
outputs, _status = cell(inputs)
```

```python
print(outputs.shape) # 모든 time-step의 hidden_state
```

```python
print(_status.shape) # (층의 개수, 배치 크기, 은닉 상태의 크기)
```

```python
# Bidirectional Recurrent Neural Network
```
